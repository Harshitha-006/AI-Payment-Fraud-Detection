# Payment fraud EDA

This notebook generates the reproducible fallback dataset, checks class imbalance, reviews distributions, and inspects numeric correlations before modeling.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import generate_synthetic_dataset

data = generate_synthetic_dataset()
data.info()
display(data['is_fraud'].value_counts(normalize=True).rename('share'))
display(data.groupby('merchant_category')['is_fraud'].mean().sort_values(ascending=False))
```

```python
numeric = data.select_dtypes('number')
sns.heatmap(numeric.corr(), cmap='vlag', center=0)
plt.title('Numeric feature correlation')
plt.show()
```
